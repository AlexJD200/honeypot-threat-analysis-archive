#### 2. Save dataframes locally for quicker query times (MUST RUN if following README guide)
The below code collects the 3 most important tables for analysis, and saves them into python pickle files for quick querying. This piece of code may take a few minutes to complete, but needs to be done now to speed up the queries in subsequent notebooks.

A lot of the code in subsequent notebooks **will not run properly** if the below code hasn't been run first. The code will create the pickle files and put them in `data/cache` - they are relatively large files, so I recommend deleting them once you have finished reviewing the notebooks.

```python
# Save dataframes to local cached files
import os

# load the main tables into pandas dataframes
print("Loading database tables into dataframes... (This may take a few minutes)")
df_sessions = pd.read_sql("SELECT * FROM sessions", engine)
df_auth = pd.read_sql("SELECT * FROM auth", engine)
df_input = pd.read_sql("SELECT * FROM input", engine)
print("Done!")

# create a cache folder if it doesn't exist
os.makedirs('../data/cache', exist_ok=True)
# save to ../data/cache
print("Saving dataframes to data/cache...")
df_sessions.to_pickle('../data/cache/sessions.pkl')
df_auth.to_pickle('../data/cache/auth.pkl')
df_input.to_pickle('../data/cache/input.pkl')
print("Done!")
```
